```python
from langchain_openai import ChatOpenAI
llm=ChatOpenAI(model="gpt-5")
```

```python
text="Hello,My name is Anmol Aditya"\
    "Email is anmoladitya0255@gmail.com"
```

```python
res=llm.invoke(f"please give me only name,email from this text:{text}")
```

```python
print(type(res.content))
```

Solution is : Strectured Output

```python
from pydantic import BaseModel,Field
from typing import List
class ResponseStrecture(BaseModel):
    name:str=Field(description="Complete Name")
    email:str=Field(description="Email Address")
    age:int=Field(description="Age")
    
strectured_llm=llm.with_structured_output(ResponseStrecture)
```

```python
res=strectured_llm.invoke(f"Please give me only name,email and age")
```

```python
res.model_dump()
```

```python
class Movies(BaseModel):
    title:str=Field(description="Movie Title")
    director:str=Field(description="Movie Director")
    year:int=Field(description="Movie Release Year")
    rating:float=Field(description="Movie Rating")

class AllMovies(BaseModel):
    movies:List[Movies]

movie_llm=llm.with_structured_output(Movies)
```

```python
movie_llm.invoke("Give me one Sci-Fi movie")
```
