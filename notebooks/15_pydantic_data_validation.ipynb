```python
from pydantic import BaseModel,Field
from typing import List ,Annotated
```

```python
class User(BaseModel):
    name:str
    age:int
```

```python
user=User(name="Anmol",age=21)


user
```

### Field validation

```python
class Product(BaseModel):
    title:str=Field(max_length=10)
    des:str=Field(min_length=20)
    price:int=Field(gt=0,le=100)
```

```python
product = Product(title="Samsung",des="This is awesome phone",price=40)
```

```python
class Order(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    product:Product
```

```python
order=Order(id="123ANM",product=product)
```

```python
order
```

```python
order.product.des
```

```python
class Carts(BaseModel):
    id:str=Field(min_lenght=6,maxlength=12)
    items:List[Product]
    comment:str=Field(default="No Comment")
```

```python
cart=Carts(id="fjgj765",items=[product])
```

### Annotated

```python
commonValidator=Annotated[List[product],Field(default=[])]
```

```python
class Carts(BaseModel):
    id:str=Field(min_lenght=6,maxlength=12)
    items:commonValidator
    comment:str=Field(default="No Comment")
```

```python
class Order(BaseModel):
    products:commonValidator
```
