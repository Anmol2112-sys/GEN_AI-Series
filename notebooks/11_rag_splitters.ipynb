### Document Text Loader

```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader=TextLoader("../data/Sql_College.txt")
texts=loader.load()
```

```python
pdfloader=PyPDFLoader("../data/System-Design-Interview-Questions.pdf")
pdfs=pdfloader.load()
```

```python
splitter=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200)
```

```python
text_chunks=splitter.split_documents(texts)
len(text_chunks)
```

```python
text_chunks
```

```python
pdf_chunks=splitter.split_documents(pdfs)
len(pdf_chunks)
```
